```python
import numpy as np
import torch

# ============================================================
# EXTRACT DATA FROM THE GRAPH
# ============================================================

X = graph.x.cpu().numpy()
y = graph.y.cpu().numpy()

train_mask = graph.train_mask.cpu().numpy().astype(bool)
val_mask = graph.val_mask.cpu().numpy().astype(bool)
test_mask = graph.test_mask.cpu().numpy().astype(bool)

# ------------------------------------------------------------
# Split features and labels
# ------------------------------------------------------------

X_train = X[train_mask]
y_train = y[train_mask]

X_val = X[val_mask]
y_val = y[val_mask]

X_test = X[test_mask]
y_test = y[test_mask]

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_val:", X_val.shape)
print("y_val:", y_val.shape)

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

# ------------------------------------------------------------
# Class distributions
# ------------------------------------------------------------

print("\nTraining class distribution:")
print(np.bincount(y_train))

print("\nValidation class distribution:")
print(np.bincount(y_val))

print("\nTest class distribution:")
print(np.bincount(y_test))
```

```python
from sklearn.model_selection import RepeatedStratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score
import numpy as np
import pandas as pd

# ============================================================
# REPEATED STRATIFIED CROSS-VALIDATION
# ============================================================

N_SPLITS = 5
N_REPEATS = 10
SEED = 42

cv = RepeatedStratifiedKFold(
    n_splits=N_SPLITS,
    n_repeats=N_REPEATS,
    random_state=SEED
)

# Use ONLY the 700 training nodes
X_cv = X_train
y_cv = y_train

# ============================================================
# DEFINE MODELS
# ============================================================

models = {

    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", LogisticRegression(
            max_iter=2000,
            random_state=SEED
        ))
    ]),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        class_weight="balanced",
        random_state=SEED,
        n_jobs=-1
    )
}

# ============================================================
# STORAGE
# ============================================================

results = []

# ============================================================
# CROSS-VALIDATION
# ============================================================

for model_name, model in models.items():

    fold_accuracies = []
    fold_macro_f1 = []

    for train_idx, val_idx in cv.split(X_cv, y_cv):

        X_fold_train = X_cv[train_idx]
        X_fold_val = X_cv[val_idx]

        y_fold_train = y_cv[train_idx]
        y_fold_val = y_cv[val_idx]

        # Train
        model.fit(X_fold_train, y_fold_train)

        # Predict
        y_pred = model.predict(X_fold_val)

        # Metrics
        accuracy = accuracy_score(
            y_fold_val,
            y_pred
        )

        macro_f1 = f1_score(
            y_fold_val,
            y_pred,
            average="macro"
        )

        fold_accuracies.append(accuracy)
        fold_macro_f1.append(macro_f1)

    # Store results
    results.append({
        "Model": model_name,

        "Accuracy Mean": np.mean(fold_accuracies),
        "Accuracy Std": np.std(fold_accuracies),

        "Macro-F1 Mean": np.mean(fold_macro_f1),
        "Macro-F1 Std": np.std(fold_macro_f1)
    })

# ============================================================
# RESULTS TABLE
# ============================================================

baseline_cv_results = pd.DataFrame(results)

print("\nREPEATED CROSS-VALIDATION RESULTS")
print("=" * 70)

print(
    baseline_cv_results.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)
```

```python
# ============================================================
# FINAL BASELINE EVALUATION ON UNTOUCHED TEST SET
# ============================================================

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    classification_report,
    confusion_matrix
)

SEED = 42

# ============================================================
# FINAL MODELS
# ============================================================

final_models = {

    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", LogisticRegression(
            max_iter=2000,
            random_state=SEED
        ))
    ]),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        class_weight="balanced",
        random_state=SEED,
        n_jobs=-1
    )
}

final_results = []

# ============================================================
# TRAIN ON ALL 700 TRAINING NODES
# ============================================================

for model_name, model in final_models.items():

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)

    macro_f1 = f1_score(
        y_test,
        y_pred,
        average="macro"
    )

    macro_precision = precision_score(
        y_test,
        y_pred,
        average="macro"
    )

    macro_recall = recall_score(
        y_test,
        y_pred,
        average="macro"
    )

    final_results.append({
        "Model": model_name,
        "Test Accuracy": accuracy,
        "Test Macro-F1": macro_f1,
        "Test Macro-Precision": macro_precision,
        "Test Macro-Recall": macro_recall
    })

    print("\n" + "=" * 70)
    print(model_name)
    print("=" * 70)

    print(f"Test Accuracy:        {accuracy:.4f}")
    print(f"Test Macro-F1:        {macro_f1:.4f}")
    print(f"Test Macro-Precision: {macro_precision:.4f}")
    print(f"Test Macro-Recall:    {macro_recall:.4f}")

    print("\nClassification Report:")
    print(
        classification_report(
            y_test,
            y_pred,
            target_names=["Low", "Medium", "High"]
        )
    )

    print("Confusion Matrix:")
    print(confusion_matrix(y_test, y_pred))


# ============================================================
# FINAL RESULTS TABLE
# ============================================================

final_baseline_results = pd.DataFrame(final_results)

print("\n\nFINAL BASELINE RESULTS")
print("=" * 70)

print(
    final_baseline_results.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import SAGEConv

# ============================================================
# PLAIN GRAPHSAGE - MEAN AGGREGATION
# ============================================================

class GraphSAGE(nn.Module):

    def __init__(
        self,
        in_channels,
        hidden_channels=64,
        out_channels=3,
        dropout=0.5
    ):
        super().__init__()

        self.conv1 = SAGEConv(
            in_channels,
            hidden_channels,
            aggr="mean"
        )

        self.conv2 = SAGEConv(
            hidden_channels,
            out_channels,
            aggr="mean"
        )

        self.dropout = dropout

    def forward(self, x, edge_index):

        x = self.conv1(x, edge_index)

        x = F.relu(x)

        x = F.dropout(
            x,
            p=self.dropout,
            training=self.training
        )

        x = self.conv2(x, edge_index)

        return x
```

```python


device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

graph = graph.to(device)

print("Device:", device)
print("Nodes:", graph.x.shape[0])
print("Features:", graph.x.shape[1])
print("Edges:", graph.edge_index.shape[1])

print("Train nodes:", graph.train_mask.sum().item())
print("Validation nodes:", graph.val_mask.sum().item())
print("Test nodes:", graph.test_mask.sum().item())
```

```python
from sklearn.metrics import accuracy_score, f1_score

# ============================================================
# GRAPHSAGE TRAINING
# ============================================================

SEED = 42

torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

model = GraphSAGE(
    in_channels=graph.x.shape[1],
    hidden_channels=64,
    out_channels=3,
    dropout=0.5
).to(device)

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001,
    weight_decay=5e-4
)

criterion = nn.CrossEntropyLoss()

# Early stopping settings
MAX_EPOCHS = 200
PATIENCE = 30

best_val_f1 = -float("inf")
best_state = None
patience_counter = 0

# ============================================================
# TRAINING LOOP
# ============================================================

for epoch in range(1, MAX_EPOCHS + 1):

    # --------------------------------------------------------
    # TRAIN
    # --------------------------------------------------------

    model.train()

    optimizer.zero_grad()

    out = model(
        graph.x,
        graph.edge_index
    )

    loss = criterion(
        out[graph.train_mask],
        graph.y[graph.train_mask]
    )

    loss.backward()

    optimizer.step()

    # --------------------------------------------------------
    # VALIDATION
    # --------------------------------------------------------

    model.eval()

    with torch.no_grad():

        val_out = model(
            graph.x,
            graph.edge_index
        )

        val_pred = val_out[
            graph.val_mask
        ].argmax(dim=1)

        val_true = graph.y[
            graph.val_mask
        ]

        val_accuracy = accuracy_score(
            val_true.cpu().numpy(),
            val_pred.cpu().numpy()
        )

        val_macro_f1 = f1_score(
            val_true.cpu().numpy(),
            val_pred.cpu().numpy(),
            average="macro"
        )

    # --------------------------------------------------------
    # EARLY STOPPING
    # --------------------------------------------------------

    if val_macro_f1 > best_val_f1:

        best_val_f1 = val_macro_f1

        best_state = {
            key: value.detach().cpu().clone()
            for key, value in model.state_dict().items()
        }

        patience_counter = 0

    else:

        patience_counter += 1



    if epoch == 1 or epoch % 10 == 0:

        print(
            f"Epoch {epoch:03d} | "
            f"Loss: {loss.item():.4f} | "
            f"Val Acc: {val_accuracy:.4f} | "
            f"Val Macro-F1: {val_macro_f1:.4f}"
        )

    if patience_counter >= PATIENCE:

        print(
            f"\nEarly stopping at epoch {epoch}"
        )

        break
```

```python

model.load_state_dict(best_state)

print(
    f"Best Validation Macro-F1: "
    f"{best_val_f1:.4f}"
)
```

```python
# ============================================================
# SYNTHETIC DATA — LOGISTIC REGRESSION
# ============================================================

import torch
import numpy as np

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    classification_report,
    confusion_matrix
)

# ------------------------------------------------------------
# 1. LOAD SYNTHETIC GRAPH
# ------------------------------------------------------------

synthetic_data = torch.load(
    "/content/drive/MyDrive/Synthetic_school_data_full.pt",
    weights_only=False
)

print("Synthetic Graph:")
print(synthetic_data)

# ------------------------------------------------------------
# 2. EXTRACT FEATURES, LABELS AND MASKS
# ------------------------------------------------------------

X_syn = synthetic_data.x.cpu().numpy()
y_syn = synthetic_data.y.cpu().numpy()

train_mask_syn = synthetic_data.train_mask.cpu().numpy()
val_mask_syn = synthetic_data.val_mask.cpu().numpy()
test_mask_syn = synthetic_data.test_mask.cpu().numpy()

# Convert masks to indices
train_idx_syn = np.where(train_mask_syn)[0]
val_idx_syn = np.where(val_mask_syn)[0]
test_idx_syn = np.where(test_mask_syn)[0]

# ------------------------------------------------------------
# 3. CREATE TRAIN / VALIDATION / TEST SETS
# ------------------------------------------------------------

X_train_syn = X_syn[train_idx_syn]
y_train_syn = y_syn[train_idx_syn]

X_val_syn = X_syn[val_idx_syn]
y_val_syn = y_syn[val_idx_syn]

X_test_syn = X_syn[test_idx_syn]
y_test_syn = y_syn[test_idx_syn]

print("\nDataset split:")
print("Training:", X_train_syn.shape, y_train_syn.shape)
print("Validation:", X_val_syn.shape, y_val_syn.shape)
print("Test:", X_test_syn.shape, y_test_syn.shape)

# ------------------------------------------------------------
# 4. TRAIN LOGISTIC REGRESSION
# ------------------------------------------------------------

lr_syn = LogisticRegression(
    max_iter=2000,
    random_state=42
)

lr_syn.fit(X_train_syn, y_train_syn)

# ------------------------------------------------------------
# 5. VALIDATION EVALUATION
# ------------------------------------------------------------

y_val_pred_syn = lr_syn.predict(X_val_syn)

val_accuracy_syn = accuracy_score(y_val_syn, y_val_pred_syn)
val_macro_f1_syn = f1_score(
    y_val_syn,
    y_val_pred_syn,
    average="macro"
)

print("\n" + "=" * 60)
print("SYNTHETIC DATA — LOGISTIC REGRESSION")
print("=" * 60)

print(f"\nValidation Accuracy: {val_accuracy_syn:.4f}")
print(f"Validation Macro-F1: {val_macro_f1_syn:.4f}")

# ------------------------------------------------------------
# 6. TEST EVALUATION
# ------------------------------------------------------------

y_test_pred_syn = lr_syn.predict(X_test_syn)

test_accuracy_syn = accuracy_score(
    y_test_syn,
    y_test_pred_syn
)

test_macro_f1_syn = f1_score(
    y_test_syn,
    y_test_pred_syn,
    average="macro"
)

test_macro_precision_syn = precision_score(
    y_test_syn,
    y_test_pred_syn,
    average="macro",
    zero_division=0
)

test_macro_recall_syn = recall_score(
    y_test_syn,
    y_test_pred_syn,
    average="macro",
    zero_division=0
)

# ------------------------------------------------------------
# 7. FINAL TEST RESULTS
# ------------------------------------------------------------

print("\nTest Accuracy:        {:.4f}".format(test_accuracy_syn))
print("Test Macro-F1:        {:.4f}".format(test_macro_f1_syn))
print("Test Macro-Precision: {:.4f}".format(test_macro_precision_syn))
print("Test Macro-Recall:    {:.4f}".format(test_macro_recall_syn))

# ------------------------------------------------------------
# 8. CLASSIFICATION REPORT
# ------------------------------------------------------------

print("\nClassification Report:")

print(
    classification_report(
        y_test_syn,
        y_test_pred_syn,
        target_names=["Low", "Medium", "High"],
        digits=2,
        zero_division=0
    )
)

# ------------------------------------------------------------
# 9. CONFUSION MATRIX
# ------------------------------------------------------------

cm_syn = confusion_matrix(
    y_test_syn,
    y_test_pred_syn
)

print("Confusion Matrix:")
print(cm_syn)
```
